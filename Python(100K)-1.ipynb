```python
condition = [
    df["Monthly_Salary"] <= 4500,
(df["Monthly_Salary"] > 4500) & (df["Monthly_Salary"] <= 5500),
df["Monthly_Salary"] > 5500
]
choices = ["Low_Salary","Medium_Salary","High_Salary"]
df["Salary_Category"] = np.select(condition, choices, default = "Unknown")
df.head(10)
```

```python
condition = [
    df["Age"] <= 30,
    (df["Age"] > 30) & (df["Age"] <= 50),
    df["Age"] > 50
]
choices = ["Young_Age", "Middle_Age","Old_Age"]
df["Age_Category"] = np.select(condition, choices, default = "Unknown")
df.head(10)
```

```python
condition = [
    df["Performance_Score"] <= 2,
    df["Performance_Score"] == 3,
    df["Performance_Score"] > 3
]
choices = ["Low_Performance", "Medium_Performance", "High_Performance"]
df["Performance_Category"] = np.select(condition, choices, default="Unknown")
df.head(10)
```

```python
condition = [
    df["Employee_Satisfaction_Score"] <= 2,
    (df["Employee_Satisfaction_Score"] > 2) & (df["Employee_Satisfaction_Score"] <= 4),
    df["Employee_Satisfaction_Score"] > 4
]
choices = ["Low_Satisfaction", "Medium_Satisfaction", "High_Satisfaction"]
df["Satisfaction_Category"] = np.select(condition, choices, default = "Unknown")
df.head(10)
```

```python
salary_analysis = df.groupby("Department").agg(
    average_salary = ("Monthly_Salary","mean"),
    maximum_salary = ("Monthly_Salary","max"),
    minimum_salary = ("Monthly_Salary","min")
).round(2).sort_values(by="average_salary",ascending=True)
print(salary_analysis)
```

```python
resign_analysis = df.groupby("Department").agg(
    employees=("Employee_ID", "count"),
    resigned_count=("Resigned", "sum"),
    resigned_rate=("Resigned", "mean")
)

# Convert to percentage
resign_analysis["resigned_rate"] = (
    resign_analysis["resigned_rate"] * 100
).round(2)

# Sort by employee count
resign_analysis = resign_analysis.sort_values(
    by="resigned_rate",
    ascending=False
)

print(resign_analysis)
```

```python
satisfaction_analysis = df.groupby("Department").agg(
    Avg_satisfaction=("Employee_Satisfaction_Score", "mean"),
    High_satisfaction=("Employee_Satisfaction_Score", "max"),
    Low_satisfaction=("Employee_Satisfaction_Score", "min")
).round(2).sort_values(by="Avg_satisfaction",ascending=False)

total_high = satisfaction_analysis["High_satisfaction"].sum()
total_low = satisfaction_analysis["Low_satisfaction"].sum()

print(satisfaction_analysis)
print("-" * 40)
print(f"Total Max Satisfaction : {total_high}")
print(f"Total Min Satisfaction : {total_low}")
```

```python
performance_analysis = df.groupby("Department").agg(
    avg_performance = ("Performance_Score","mean"),
    high_performance = ("Performance_Score","max"),
    low_performance = ("Performance_Score","min")
).round(2)
total_high_per = performance_analysis["high_performance"].sum()
total_low_per = performance_analysis["low_performance"].sum()
print(performance_analysis)
print('-'*40)
print(f"Total High Performer : {total_high_per}")
print(f"Total Low Performer : {total_low_per}")
```

```python
Working_Analysis = df.groupby("Department").agg(
    avg_working_hr = ("Work_Hours_Per_Week","mean"),
    high_work_hrs = ("Work_Hours_Per_Week","max"),
    low_work_hrs = ("Work_Hours_Per_Week","min"),
    total_hrs = ("Work_Hours_Per_Week","sum") 
).round(2).sort_values(by="avg_working_hr",ascending=False)
total_max_hrs = Working_Analysis["high_work_hrs"].sum()
total_min_hrs = Working_Analysis["low_work_hrs"].sum()
print(Working_Analysis)
print('-'*40)
print(f"Total Maximum Working Hours : {total_max_hrs}")
print(f"Total Minimum Working Hours : {total_min_hrs}")
```
